```python
import numpy as np
import matplotlib.pyplot as plt


def ideal_capacity_curve(task_complexity: np.ndarray) -> np.ndarray:
    """
    Conceptual mapping from task complexity to required LoRA capacity.

    This function is illustrative rather than empirically fitted.
    """
    return 3 * np.sqrt(task_complexity) + 1


def plot_appropriate_allocation(output_path: str) -> None:
    x = np.linspace(0.5, 10, 200)
    y = ideal_capacity_curve(x)

    client_x = np.array([1.2, 2.5, 4.2, 6.3, 8.7])
    ideal_y = ideal_capacity_curve(client_x)

    fig, ax = plt.subplots(figsize=(4.2, 5.4))

    ax.plot(
        x,
        y,
        linewidth=2,
        label="Ideal capacity curve",
    )

    ax.scatter(
        client_x,
        ideal_y,
        s=45,
        marker="o",
        label="Assigned rank",
        zorder=3,
    )

    for i, (px, py) in enumerate(zip(client_x, ideal_y), start=1):
        ax.text(
            px + 0.12,
            py + 0.10,
            f"C{i}",
            fontsize=8,
        )

    ax.set_xlim(0, 10.5)
    ax.set_ylim(0, 11.5)

    ax.set_xlabel("Task Complexity")
    ax.set_ylabel("Required Model Capacity\n(LoRA Rank)")
    ax.set_title("(b) Appropriate Rank Allocation")

    ax.grid(
        True,
        linestyle="--",
        alpha=0.4,
    )

    ax.legend(
        frameon=False,
        loc="upper left",
    )

    fig.tight_layout()
    fig.savefig(
        output_path,
        dpi=300,
        bbox_inches="tight",
    )
    plt.close(fig)


def plot_rank_misallocation(output_path: str) -> None:
    x = np.linspace(0.5, 10, 200)
    y = ideal_capacity_curve(x)

    client_x = np.array([1.2, 2.5, 4.2, 6.3, 8.7])
    ideal_y = ideal_capacity_curve(client_x)

    # Conceptual allocation errors:
    # positive deviation = excessive rank;
    # negative deviation = insufficient rank.
    assigned_y = np.array([
        ideal_y[0] + 2.0,
        ideal_y[1] - 1.6,
        ideal_y[2] + 2.2,
        ideal_y[3] - 2.0,
        ideal_y[4] - 1.5,
    ])

    fig, ax = plt.subplots(figsize=(4.2, 5.4))

    ax.plot(
        x,
        y,
        linewidth=2,
        label="Ideal capacity curve",
    )

    ax.scatter(
        client_x,
        ideal_y,
        s=45,
        marker="o",
        label="Ideal rank",
        zorder=3,
    )

    ax.scatter(
        client_x,
        assigned_y,
        s=45,
        marker="x",
        linewidths=2,
        label="Assigned rank",
        zorder=4,
    )

    for i, (px, ideal_rank, assigned_rank) in enumerate(
        zip(client_x, ideal_y, assigned_y),
        start=1,
    ):
        ax.plot(
            [px, px],
            [ideal_rank, assigned_rank],
            linestyle="--",
            linewidth=1,
        )

        ax.text(
            px + 0.12,
            assigned_rank + 0.12,
            f"C{i}",
            fontsize=8,
        )

    ax.annotate(
        "Excess capacity",
        xy=(client_x[0], assigned_y[0]),
        xytext=(2.8, 10.2),
        arrowprops={
            "arrowstyle": "->",
        },
        fontsize=8,
    )

    ax.annotate(
        "Capacity shortage",
        xy=(client_x[3], assigned_y[3]),
        xytext=(6.5, 2.2),
        arrowprops={
            "arrowstyle": "->",
        },
        fontsize=8,
    )

    ax.set_xlim(0, 10.5)
    ax.set_ylim(0, 11.5)

    ax.set_xlabel("Task Complexity")
    ax.set_ylabel("Required Model Capacity\n(LoRA Rank)")
    ax.set_title("(c) Rank Misallocation")

    ax.grid(
        True,
        linestyle="--",
        alpha=0.4,
    )

    ax.legend(
        frameon=False,
        loc="upper left",
    )

    fig.tight_layout()
    fig.savefig(
        output_path,
        dpi=300,
        bbox_inches="tight",
    )
    plt.close(fig)


if __name__ == "__main__":
    plot_appropriate_allocation(
        "appropriate_rank_allocation_curve.png"
    )

    plot_rank_misallocation(
        "rank_misallocation_curve.png"
    )
```
